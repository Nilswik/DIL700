```python
import tensorflow as tf


# Create two sample tensors 
tensor_a = tf.constant([[1, 2], [3, 4]], dtype=tf.float32)
tensor_b = tf.constant([[5, 6], [7, 8]], dtype=tf.float32)
# Perform element-wise addition
tensor_c = tf.add(tensor_a, tensor_b)
print("Result of element-wise addition:\n", tensor_c.numpy())
```

```python
import matplotlib.pyplot as plt

plt.rc('font', size=14)
plt.rc('axes', labelsize=14, titlesize=14)
plt.rc('legend', fontsize=14)
plt.rc('xtick', labelsize=10)
plt.rc('ytick', labelsize=10)
```

```python
# Element-wise multiplication
tensor_d = tf.multiply(tensor_a, tensor_b)
print("Result of element-wise multiplication:\n", tensor_d.numpy())
```

```python
#Reshape a tensor 
tensor_e = tf.reshape(tensor_a, [4, 1])
tensor_f = tf.reshape(tensor_b, [1, 4])
print("Reshaped tensor:\n", tensor_e.numpy())
print("Reshaped tensor:\n", tensor_f.numpy())
```

```python
# Slicing and indexing tensor 
tensor_g = tensor_a[1:, :]
print("Sliced tensor:\n", tensor_g.numpy())

tensor_h = tensor_b[:, 0]
print("Indexed tensor:\n", tensor_h.numpy())

```

```python
# combine tensors 
tensor_i = tf.concat([tensor_a, tensor_b], axis=0)
print("Concatenated tensor along axis 0:\n", tensor_i.numpy())

# Combine tensors along y- axis 
tensor_j = tf.stack([tensor_a, tensor_b], axis=1)
print("Stacked tensor along axis 1:\n", tensor_j.numpy())
```

```python
# Splitting tensor 
tensor_k = tf.split(tensor_i, num_or_size_splits=2, axis=0) # use combined tensor_i
print("Splitted tensors:")
for idx, t in enumerate(tensor_k):
    print(f"Tensor {idx}:\n", t.numpy())    
```

```python
# Basic math operations on tensors 
tensor_sum = tf.reduce_sum(tensor_a)
print("Sum of all elements in tensor_a:", tensor_sum.numpy())

tensor_mean = tf.reduce_mean(tensor_b)
print("Mean of all elements in tensor_b:", tensor_mean.numpy())

tensor_max = tf.reduce_max(tensor_a)
print("Max element in tensor_a:", tensor_max.numpy())

# Transpose a tensor
tensor_l = tf.transpose(tensor_b)
print("Transposed tensor_b:\n", tensor_l.numpy())

# Reshape back to original shape
tensor_m = tf.reshape(tensor_l, [2, 2])
print("Reshaped back to original shape:\n", tensor_m.numpy())

# Expand dimensions
tensor_n = tf.expand_dims(tensor_a, axis=0)
print("Expanded dimensions of tensor_a:\n", tensor_n.numpy())

# Squeeze dimensions
tensor_o = tf.squeeze(tensor_n)
print("Squeezed tensor back to original shape:\n", tensor_o.numpy())

# Create a tensor filled with ones
tensor_p = tf.ones([5, 5], dtype=tf.float32)
print("Tensor filled with ones:\n", tensor_p.numpy())

# Create a tensor filled with zeros
tensor_q = tf.zeros([3, 3], dtype=tf.float32)
print("Tensor filled with zeros:\n", tensor_q.numpy())

# Create a tensor with random values
tensor_r = tf.random.uniform([2, 2], minval=0, maxval=10, dtype=tf.float32)
print("Tensor with random values:\n", tensor_r.numpy())

```

```python
# Neural netwrok Foundations

# Define a simple neural netwrok model using TensorFlow/Keras

model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(32,)),
    tf.keras.layers.Dense(64, activation='relu'),
    tf.keras.layers.Dense(10, activation='softmax')
])


# Compile the model
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

# Print model summary

model.summary()
```

```python

# Load MNIST dataset
mnist = tf.keras.datasets.mnist
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Flatten and normalise
x_train = x_train.reshape(-1, 28 * 28).astype("float32") / 255.0
x_test = x_test.reshape(-1, 28 * 28).astype("float32") / 255.0
```

```python
mnist_model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(784,)),
    tf.keras.layers.Dense(64, activation="relu"),
    tf.keras.layers.Dense(10, activation="softmax")
])

mnist_model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

```

```python
history_mnist = mnist_model.fit(
    x_train,
    y_train,
    epochs=10,
    batch_size=32,
    validation_split=0.2,
    verbose=1
)
```

```python
test_loss_mnist, test_acc_mnist = mnist_model.evaluate(x_test, y_test, verbose=0)
print(f"MNIST Test Loss: {test_loss_mnist:.4f}")
print(f"MNIST Test Accuracy: {test_acc_mnist:.4f}")
```

```python
plt.figure()
plt.plot(history_mnist.history["loss"])
plt.plot(history_mnist.history["val_loss"])
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.title("MNIST Training and Validation Loss")
plt.legend(["Training Loss", "Validation Loss"])
plt.show()
```

```python
plt.figure()
plt.plot(history_mnist.history["accuracy"])
plt.plot(history_mnist.history["val_accuracy"])
plt.xlabel("Epochs")
plt.ylabel("Accuracy")
plt.title("MNIST Training and Validation Accuracy")
plt.legend(["Training Accuracy", "Validation Accuracy"])
plt.show()
```

```python
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences

# Load IMDb dataset
num_words = 20000
(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words=num_words)

# Pad sequences
max_len = 256
x_train = pad_sequences(x_train, maxlen=max_len)
x_test = pad_sequences(x_test, maxlen=max_len)
```

```python
# imdb_model = tf.keras.Sequential([
#     tf.keras.layers.Embedding(num_words, 128),
#     #tf.keras.layers.Flatten(), # prevent overfiting
#     tf.keras.layers.GlobalAveragePooling1D(),
#     tf.keras.layers.Dense(64, activation="relu"),
#     tf.keras.layers.Dense(1, activation="sigmoid")
# ])

imdb_model = tf.keras.Sequential([
    tf.keras.layers.Embedding(
        input_dim=num_words,
        output_dim=128
    ),

    # Global average pooling to reduce overfitting
    tf.keras.layers.GlobalAveragePooling1D(),

    # Regularised dense layer
    tf.keras.layers.Dense(
        64,
        activation="relu",
        kernel_regularizer=tf.keras.regularizers.l2(0.01)
    ),

    # Dropout for further regularisation
    tf.keras.layers.Dropout(0.5),

    # Binary classification output
    tf.keras.layers.Dense(1, activation="sigmoid")
])
```

```python
imdb_model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
```

```python
history_imdb = imdb_model.fit(
    x_train,
    y_train,
    epochs=10,
    batch_size=32,
    validation_split=0.2,
    verbose=1
)
```

```python
test_loss_imdb, test_acc_imdb = imdb_model.evaluate(x_test, y_test, verbose=0)
print(f"IMDb Test Loss: {test_loss_imdb:.4f}")
print(f"IMDb Test Accuracy: {test_acc_imdb:.4f}")
```

```python
plt.figure()
plt.plot(history_imdb.history["loss"])
plt.plot(history_imdb.history["val_loss"])
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.title("IMDb Training and Validation Loss")
plt.legend(["Training Loss", "Validation Loss"])
plt.show()
```

```python
plt.figure()
plt.plot(history_imdb.history["accuracy"])
plt.plot(history_imdb.history["val_accuracy"])
plt.xlabel("Epochs")
plt.ylabel("Accuracy")
plt.title("IMDb Training and Validation Accuracy")
plt.legend(["Training Accuracy", "Validation Accuracy"])
plt.show()
```
