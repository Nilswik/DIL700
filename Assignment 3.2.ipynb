```python
from tensorflow import keras
import matplotlib.pyplot as plt
import numpy as np
import datetime
import os
import tensorflow as tf


np.random.seed(42)

(x_train, y_train), (x_test, y_test) = keras.datasets.fashion_mnist.load_data() # Load Fashion-MNIST dataset /784 dimensions

X_train = x_train.reshape(-1, 28*28).astype(np.float32) / 255.0 # Normalizing the data
X_test  = x_test.reshape(-1, 28*28).astype(np.float32) / 255.0

print(X_train.shape, X_test.shape)

lbls = {0: 'T-shirt/top', 1: 'Trouser', 2: 'Pullover', 3: 'Dress', 4: 'Coat',
        5: 'Sandal', 6: 'Shirt', 7: 'Sneaker', 8: 'Bag', 9: 'Ankle boot'}

plt.figure(figsize=(10,4))
for i in range(10):
    plt.subplot(2, 5, i+1)
    plt.imshow(x_train[i], cmap='gray')
    plt.axis('off')
    plt.title(f'Label: {lbls[y_train[i]]}')
```

```python
activation = 'relu'
#activation = 'elu'
#activation = 'selu'

model = keras.Sequential([
    keras.layers.InputLayer(input_shape=(28*28,)),
    *[keras.layers.Dense(100, activation=activation) for i in range(100)],
    keras.layers.Dense(10, activation='softmax')
])

#model.summary()

base_dir = r"C:\Users\nisse\Skola\DIL700\Assignment3.2"

log_dir = os.path.join(
    base_dir, "logs", "fit",
    f"activation_{activation}",
    datetime.datetime.now().strftime("%Y%m%d-%H%M%S")
)

tb_callback = tf.keras.callbacks.TensorBoard(
log_dir=log_dir,
histogram_freq=1)

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
                metrics=['accuracy'])
model.fit(X_train, y_train, epochs=10, batch_size=32, callbacks = [tb_callback], validation_split=0.2)
```

```python
#activation = 'relu'
activation = 'elu'
#activation = 'selu'

keras.backend.clear_session()

model = keras.Sequential([
    keras.layers.InputLayer(input_shape=(28*28,)),
    *[keras.layers.Dense(100, activation=activation) for i in range(100)],
    keras.layers.Dense(10, activation='softmax')
])

#model.summary()

base_dir = r"C:\Users\nisse\Skola\DIL700\Assignment3.2"

log_dir = os.path.join(
    base_dir, "logs", "fit",
    f"activation_{activation}",
    datetime.datetime.now().strftime("%Y%m%d-%H%M%S")
)

tb_callback = tf.keras.callbacks.TensorBoard(
log_dir=log_dir,
histogram_freq=1)

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
                metrics=['accuracy'])
model.fit(X_train, y_train, epochs=10, batch_size=32, callbacks = [tb_callback], validation_split=0.2)
```

```python
#activation = 'relu'
#activation = 'elu'
activation = 'selu'

keras.backend.clear_session()

model = keras.Sequential([
    keras.layers.InputLayer(input_shape=(28*28,)),
    *[keras.layers.Dense(100, activation=activation) for i in range(100)],
    keras.layers.Dense(10, activation='softmax')
])

#model.summary()

base_dir = r"C:\Users\nisse\Skola\DIL700\Assignment3.2"

log_dir = os.path.join(
    base_dir, "logs", "fit",
    f"activation_{activation}",
    datetime.datetime.now().strftime("%Y%m%d-%H%M%S")
)

tb_callback = tf.keras.callbacks.TensorBoard(
log_dir=log_dir,
histogram_freq=1)

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
                metrics=['accuracy'])
model.fit(X_train, y_train, epochs=10, batch_size=32, callbacks = [tb_callback], validation_split=0.2)
```

Not one of the activation models ever perform over 11% accuracy indicating that the selection is simply random. The relu function probably suffer from the dying relu problem. Negative inputs are set to zero which can be an advantage, but in this case it probably means that after some layers the gradient is gone. in the bias histogram we can see a clear peak very close to zero indicating that the gradient dies out. from the kernel histogram we see that the distrubution of the weights in the layers remain more or less the same indicating that the model is not learning anything 

The elu activation function has the ability to set negative values to small negative values. This means that the gradients dont die out in the same way as in relu making it benificial when using very deep neural networks. the bias histogram does show a wider distrubution of the weigths, but most are still close to zero. The kernel histogram shows that something actually happens and the model is trying to learn. This model does too die out however.



The selu activation function shows a wider distrubution of weights in the bias histogram. The kernel histogram is more consistent then the elu histogram. This is probably due to the self normalizing characteristics of the activation function. The accuracy is still low though. in https://www.geeksforgeeks.org/deep-learning/selu-activation-function-in-neural-network/ it is mentioned that an alpha dropout should be used. Maybe that is why.
