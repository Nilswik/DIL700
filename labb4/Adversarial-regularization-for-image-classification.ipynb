Adversarial learning revolves around training a model with both normal training data and data that has been intentionally altered (adversarial examples). Despite appearing normal to the human eye, these adversarial examples are specifically designed to trick the model into making incorrect predictions or classifications. The goal of adversarial training is to help the model become more resilient to such adversarial perturbations, so that it can accurately predict even when faced with such examples. By being exposed to these intentionally misleading examples during training, the model learns to make more robust predictions.
